```python
from dotenv import load_dotenv 

load_dotenv()
```

```python
from langchain.chat_models.base import init_chat_model

llm = init_chat_model(model="gpt-4o-mini", model_provider="openai")
```

```python
from langchain_core.tools import tool 

# @tool: 데코레이터를 사용하여 도구 정의
# add: 도구의 이름
# description: 도구에 대한 설명
# a, b: 도구가 받을 매개변수
# return: 도구가 반환하는 값의 타입(Expected Argument)
@tool 
def add(a: int, b: int) -> int:
    """숫자 a와 b를 더합니다."""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """숫자 a와 b를 곱합니다."""
    return a * b
```

```python
llm_with_tools = llm.bind_tools([add, multiply])
```

```python
query = "3 곱하기 5는 무엇인가요?"
```

```python
llm.invoke(query)
```

```python
result = llm_with_tools.invoke(query)
```

```python
result.tool_calls
```

```python
from langchain_core.messages import AnyMessage, HumanMessage 
from typing import Sequence 

human_message = HumanMessage(query)
message_list: Sequence[AnyMessage] = [human_message]
```

```python
ai_message = llm_with_tools.invoke(message_list)
```

```python
ai_message
```

```python
ai_message.tool_calls
```

```python
message_list.append(ai_message)
```

```python
multiply.invoke(ai_message.tool_calls[0])
```

```python
tool_message = multiply.invoke(ai_message.tool_calls[0])
```

```python
message_list.append(tool_message)
```

```python
llm_with_tools.invoke(message_list)
```
